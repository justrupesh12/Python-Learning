```python
# A python program to display from a string

str="Hello"
for i in str:
    print(i)
```

```python
# A python program yo display each character from string using sequence index

str='Hello'
n=len(str)

for i in range(n):
    print(str[i])
    
```

```python
#A program a display the elements of a list using for loop.
list=[10,20.5,'A','America']
for element in list:
    print(element)
```

```python
# A python  programm a display and find the Sum of a list of numbers using for loop 
list=[10,20,30,40,50]
sum=0
for i in list:
    print(i)
    sum=sum+i
print('sum= ', sum)    
```

```python
# A python  programm a display and find the Sum of a list of numbers using while loop 
list=[10,20,30,40,50]
sum=0
i=0
while i<len(list):
    print(list[i])
    sum=sum+list[i]
    i=i+1

print('sum', sum)
```

```python
# A py program to display the right angle traingle from using nested loop
for i in range(1,11):
    for j in range(1,i+1):
        print("*",end=" ")
    print()    
        
    
```

```python
# a python program to display * in a equivalent triangle

n= int(input("please enter the number to print tringle"))

for i in range(1,11):
    print(" " * n,end="")
    print('* ' * i)
    n=n-1
          
    
```
